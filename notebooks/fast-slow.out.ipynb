#### Slow Start and Fast Start Scenarios

##### Slow Start

In the context of FIA, the “slow start” scenarios we simulated are intended to represent a budgetary or workforce shortfall in the first $z$ years early in an inventory cycle, necessitating a planned reduction of field visitation by a fixed proportion of micropanels per macropanel in the first $z$ macropanels in $\mathcal{P}_M$.

Then, a budgetary or workforce recovery occurs, enabling a planned increase of field visitation by the same fixed proportion of micropanels per macropanel in the subsequent $z$ macropanels. Finally, work capacity stabilizes, and the $k$ micropanels in the final macropanel $P_M$ in $\mathcal{P}_M$ retain their original assignments (i.e., we’re “back on time”).

##### Fast Start

Under the “fast start” scenarios we simulated, our VAE intentions are analogous to the slow start scenario described above, but the timing of the availability of resources of field work is reversed. That is, extra resources for field work enable a fixed proportion of “early” micropanels per macropanel for the first $z$ macropanels in $\mathcal{P}_M$. Annual sampling effort is then reduced by the same proportion micropanels per macropanel for the subsequent $z$ macropanels to  
“get back on schedule”; and the nominally prescribed rate of $k$ micropanels per macropanel is reestablished for macropanel $M$. A more detailed demonstrative example can be found in @sec-supp-mat.

##### Generalizing Shifted Macropanel Boundaries

If we preserve $R$ and $M$, but modify the original macropanel assignments of some micropanels for VAE purposes as the slow and fast start scenarios are intended to represent, the micropanels no longer evenly partition the macropanels throughout $\mathcal{P}_M$. Therefore, it is convenient to extend our notation to represent an arbitraty, newly constituted cycle $\mathcal{P}^\prime_M$ following repositioning of micropanels (i.e., to represent and facilitate the application of a particular slow or fast start VAE regimen across any value of $M \in \{5, 7, 10, 14\}$).

Let $\mathbf{U}_M$ be the baseline uniform boundary sequence vector of length $M+1$, where the elements are determined by the constant step-size $k = \frac{R}{M}$:

$$\mathbf{U}_M = (u_0, u_1, \dots, u_M), \quad \text{where } u_m = m \cdot k$$

To then specify a slow or fast start regimen that can be applied arbitrarily for any value of $M$, we further define the boundary deviation vector $\boldsymbol{\delta}_M$ of length $M+1$ as:

$$\boldsymbol{\delta}_M = (\delta_0, \delta_1, \ldots, \delta_M)$$

The shifted boundary sequence vector $\mathbf{B}$ can then be obtained by summing the baseline boundary vector and the boundary deviation vector:

$$\mathbf{B} = \mathbf{U}_M + \boldsymbol{\delta}_M = (u_0 + \delta_0, u_1 + \delta_1, \dots, u_M + \delta_M)$$

If the originally prescribed boundaries are unshifted, $\boldsymbol{\delta}_M$ is an $M + 1$ vector of $0$’s, and $\mathbf{B}$ is therefore simply the baseline boundary vector $\mathbf{U}_M$ under the originally prescribed distribution of micropanels across macropanels.

$\mathbf{B}$ then takes the form

$$\mathbf{B} = (b_0, b_1, \ldots,b_M),$$ where the $M+1$ elements therein represent an intitial index $b_0$ and the $M$ new upper micropanel boundaries; i.e., the values of $r$ associated with the last of each  
of the sequence of shifted micropanels in the newly reconstituted macropanels. Each modified macropanel $P_m*$ ($m* = 1,\ldots, M$) can then be constructed by obtaining indices for micropanels from between adjacent elements of $\mathbf{B}$:

$$P_m* = \{ q_r \in \mathcal{Q} : b_{m-1} + 1 \leq r \leq b_m \}.$$

##### Modified Macropanel Lookup Function

Because the distributions of micropanels across macropanels are no longer uniform under these shifted configurations, the scalar ceiling function cannot be used to determine macropanel assignment $m$ of any macropanel $r$.  
However, such a function is still needed to describe and implement the slow start and fast start VAE scenarios.

We can define a modified macropanel lookup function $f_b(r)$ that counts how many upper bounds $b_m$ from the sequence $\mathbf{B}$ are strictly less than $r$, and then offsets the result by 1. Using bracket notation (where the brackets evaluate to $1$ if the enclosed condition is true, and $0$ if false):

$$f_b(r) = 1 + \sum_{m=1}^{M-1} [r > b_m]$$

Demonstrative examples of the construction of $\mathbf{U}_M$, $\boldsymbol{\delta}_M$, $\mathbf{B}$, and the use of $f_b(r)$ can be found in @sec-supp-mat.
