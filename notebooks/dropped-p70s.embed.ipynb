#### Dropped Micropanels

Under the slow and fast start scenarios, both the total number of micropanels $R$ and the total number of macropanels $M$ within the cycle are preserved, but some micropanels are reassigned to other macropanels such that the number of micropanels per macropanel is no longer the fixed value $k$ as under the original configuration. Under the extended and shortened cycle length scenarios, the total number of micropanels $R$ is retained, but micropanels are evenly and sequentially redistributed across macropanels such that $M$ either gets larger or smaller.

It’s conceivable that FIA could encounter such dire obstacles to field operations that sampling intensities must be consistently reduced within every macropanel in a cycle, with no intention (or capability) to regain the nominal sampling intensity before the end of the cycle. This could be achieved by simply eliminating one or more micropanels per macropanel from consideration for field visitation, until its prescribed position in the following inventory cycle at the earliest. We refer to this cirucmstance as a “dropped micropanel” scenario.

##### Micropanel Exclusion Notation

To generalize dropped micropanel scenarios, let $d \in \{0, 1, \dots, k\}$ denote the number of trailing micropanels dropped from each macropanel in a cycle. Let $\mathcal{D}_{M,d} \subset \mathcal{Q}$ be the set of dropped micropanels for cycle configuration $\mathcal{P}_M$ under a sampling intensity reduction of the last $d$ micropanels in each macropanl. Because the final $d$ micropanels in macropanel $m$ span the micropanel indices $r = mk - d + 1$ through $r = mk$, the set of dropped micropanels can be expressed as:

$$\mathcal{D}_{M,d} = \{ q_r \in \mathcal{Q} : mk - d + 1 \leq r \leq mk, \text{ for } m = 1, 2, \dots, M \}$$

The modified cycle $\mathcal{P}'_{M,d}$ preserves the original $M$ distinct macropanels, but truncates the upper micropanel boundary within each macropanel at $r = mk - d$:

$$\mathcal{P}'_{M,d} = \{P'_1, P'_2, \dots, P'_M\}$$

where each truncated macropanel $P'_m$ (for $m = 1, 2, \dots, M$) is explicitly defined by:

$$P'_m = \{ q_r \in \mathcal{Q} : (m-1)k + 1 \leq r \leq mk - d \}$$

Under this generalized form, the modified number of micropanels per macropanel is reduced from $k$ to $k - d$. Note that when $d = 0$, $\mathcal{D}_{M,0} = \emptyset$ and $\mathcal{P}'_{M,0} = \mathcal{P}_M$ (because the inequality in the expression for $\mathcal{D}_{M,d}$ yields no valid integer solution); when $d = k$, all micropanels are dropped ($\mathcal{D}_{M,k} = \mathcal{Q}$) and each active macropanel becomes empty ($P'_m = \emptyset$).

In the simulation study, we evaluated $d = (1,2,3)$ for each of the three states.
