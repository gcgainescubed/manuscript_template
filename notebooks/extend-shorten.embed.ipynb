#### Extended and Shortened Cycle Lengths

Under the slow and fast start VAE scenarios described above, the number of macropanels in the cycle $\mathcal{P}_M$ (i.e., $M$, which we can think of as corresponding to “inventory cycle length” in the context of FIA) is preserved by shifting an equivalent number of micropanels forward (for delayed field visitation) in some macropanels as are held backward (for early field visitation) in others.

Another way VAE might be applied, for particular values of $R$ and $M$, is to evenly redistribute micropanels into a modified total number of macropanels, say $M^{\prime} \in \{5,7,10,14\}, \quad M^{\prime} \neq M$, where $M^{\prime}$ is still a factor of $R$. If $M^{\prime} > M$ the cycle length is extended, and the macropanel sampling intensity $k$ (i.e., the annual effort) is reduced. Conversely, if $M^{\prime} < M$ the cycle length is shortened and the annual effort in each newly configured macropanel $P^\prime_m$ in in the modified cycle $\mathcal{P}^\prime_M$ is increased.

##### A modified macropanel lookup function for extended and shortened cycle lengths

To map any micropanel from its originally prescribed macropanel $P_m$ to its new macropanel $P^\prime_m$ under a shortened or extended cycle length, we define a function that uses both the original macropanel index $m$ and the local micropanel offset $l$ within that macropanel.

Let:

- $M, k$ be the parameters of the original configuration (e.g., $M=5, k=14$).
- $M^\prime, k^\prime$ be the parameters of the new configuration (e.g., $M^\prime=7, k^\prime=10$).
- $m$ be the original macropanel index, where $m \in \{1, 2, \dots, M\}$.
- $l$ be the local index of the micropanel within $P_m$, where $l \in \{1, 2, \dots, k\}$.

A micropanel’s local macropanel coordinate (i.e., the two unique values that determine its chronological position within its originally assigned macropanel and cycle) $(m,l)$ maps to its static global index $r$ via:

$$r = (m - 1)k + l$$

In the new cycle configuration $\mathcal{P}^\prime_M$, each macropanel $P^\prime_m$ contains exactly $k^\prime$ micropanels. To determine which macropanel (i.e., which sequence of $k^\prime$ micropanels) the index $r$ falls within, we divide $r$ by $k^\prime$ and apply the ceiling function $\lceil \cdot \rceil$ to round up to the nearest integer:

$$m^\prime = \left\lceil \frac{r}{k^\prime} \right\rceil$$

Combining both steps yields the definitive closed-form mapping function $\mathbf{f}(m,l)$:

$$f^\prime(m,l) = \left\lceil \frac{(m - 1) k + l}{k^\prime} \right\rceil$$

Where $\mathbf{f}(m,l) = m^\prime$ represents the index of the modified macropanel ($m^\prime \in \{1, 2, \dots, M^\prime\}$) to which micropanel $r$ is newly assigned and $\lceil \cdot \rceil$ represents the ceiling function described above.
