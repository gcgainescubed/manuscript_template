##### Modified Partition Notation (Shifted Configurations)

For an uneven or shifted layout where subpanels are structurally rearranged, the uniform mapping breaks down. To formalize this, we introduce an explicit boundary sequence vector $\mathbf{B}$ containing $M+1$ elements:

$$\mathbf{B} = (b_0, b_1, \dots, b_M)$$

For the specific modified configuration where $M = 5$, the boundaries are explicitly assigned as:

$$\mathbf{B} = (0, 10, 20, 38, 56, 70)$$

Each individual macro panel $P_i$ (for $i = 1, 2, \dots, M$) is then constructed by pulling indices between adjacent elements of the boundary sequence:

$$P_i = \{ q_j \in \mathcal{Q} \mid b_{i-1} + 1 \leq j \leq b_i \}$$

##### Explicit Macro Panel Contents

- $P_1 = \{ q_j \in \mathcal{Q} \mid 1 \leq j \leq 10 \}$
- $P_2 = \{ q_j \in \mathcal{Q} \mid 11 \leq j \leq 20 \}$
- $P_3 = \{ q_j \in \mathcal{Q} \mid 21 \leq j \leq 38 \}$
- $P_4 = \{ q_j \in \mathcal{Q} \mid 39 \leq j \leq 56 \}$
- $P_5 = \{ q_j \in \mathcal{Q} \mid 57 \leq j \leq 70 \}$

##### Modified Panel Lookup Function Alternatives

Because the subpanel distributions are no longer uniform, the scalar ceiling function cannot be used. Below are two equivalent mathematical approaches to express the lookup function $f(j)$ for the shifted configuration.

##### Approach 1: The Summation / Indicator Function Approach

We can define the lookup function $f(j)$ by counting how many boundaries the subpanel index $j$ has successfully crossed. Using the Iverson bracket notation (where the brackets evaluate to $1$ if the enclosed condition is true, and $0$ if false), the macro panel index can be calculated directly by checking membership across all intervals:

$$f(j) = \sum_{i=1}^{M} i \cdot [b_{i-1} + 1 \leq j \leq b_i]$$

Alternatively, an even more concise formulation counts how many upper bounds $b_i$ from our sequence $\mathbf{B} = (0, 10, 20, 38, 56, 70)$ are strictly less than $j$, then offsets the result by 1:

$$f(j) = 1 + \sum_{i=1}^{M-1} [j > b_i]$$

##### Example Evaluation

To locate subpanel $q_{25}$ using the upper bounds $b_1=10, b_2=20, b_3=38, b_4=56$:

$$f(25) = 1 + [25 > 10] + [25 > 20] + [25 > 38] + [25 > 56]$$ $$f(25) = 1 + 1 + 1 + 0 + 0 = 3$$

This correctly maps subpanel $q_{25}$ to Macro Panel $P_3$.
