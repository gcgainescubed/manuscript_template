### Initial Notation

#### Sample locations

Let

$$\mathbf{s}_i = \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$$ be the bivariate spatial coordinate vector for the $i^{th}$ FIA sample location in a focal US state, $i = 1,\ldots,n$, where is $n$ is the number of all sample locations in the focal state, and $x_1$ and $x_2$ are measures of longitude and latitude respectively.

#### Micropanels

Let

$$\mathcal{Q} = \{q_r \vert r=1, \ldots, R\}$$

be a set of “micropanels” $q_r$ that partition the $n$ sample locations into approximately equivalently-sized subsets, such that each $q_r \in \mathcal{Q}$ achieves approximate spatial dispersion across the focal state of each of the $s_i$ it contains. The $q_r$ defined here are meant to correspond to collections of sample locations referred to in the FIA Program as “Panel 70’s”, which imply $R = 70$, or “Panel 140’s”, which imply $R = 140$.

In CONUS states where the FIA Program is administered by the Pacific Northwest Research Station, $R = 140$ in present day inventory operations. In the rest of the CONUS states, $R = 70$. Though this notation is intended to be flexible enough to accommodate a range of possible values of $R$, in this study we don’t consider PNWRS states, and so we’ll only consider $R = 70$ for practical purposes.

#### Layout Parameters

The 70 subpanels are partitioned into $M$ macro panels based on the active configuration choice:

$$M \in \{5, 7, 10, 14\}$$

Because the subpanels are evenly distributed, the number of subpanels grouped into each macro panel, denoted as $k$, is strictly determined by the choice of $M$:

$$k = \frac{70}{M}$$

#### The Partition Notation

For any chosen configuration $M$, the system layout partition $\mathcal{P}_M$ is a set of $M$ distinct macro panels:

$$\mathcal{P}_M = \{P_1, P_2, \dots, P_M\}$$

Where each individual macro panel $P_i$ (for $i = 1, 2, \dots, M$) is a collection of subpanels defined by their static global identity indices:

$$P_i = \{ q_j \in \mathcal{Q} \mid (i-1)k + 1 \leq j \leq ik \}$$

#### Panel Lookup Function

To map any given subpanel global index $j$ directly to its corresponding macro panel index $i$ under configuration $M$, we define the lookup function using the ceiling function $\lceil \cdot \rceil$:

$$f(j) = \left\lceil \frac{j}{k} \right\rceil = \left\lceil \frac{j \cdot M}{70} \right\rceil$$

Where:

- $j$ is the global index of the subpanel ($1 \leq j \leq 70$).
- $M$ is the number of active macro panels ($M \in \{5, 7, 10, 14\}$).
- $f(j) = i$ outputs the macro panel index ($1 \leq i \leq M$).
