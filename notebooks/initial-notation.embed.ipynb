### Initial Notation

#### Sample locations

Let

$$\mathbf{s}_i = \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}$$ be the bivariate spatial coordinate vector for the $i^{th}$ FIA sample location in a focal US state, $i = 1,\ldots,n$, where $n$ is the number of all sample locations in the focal state, and $x_1$ and $x_2$ are measures of longitude and latitude respectively.

#### Micropanels

Let

$$\mathcal{Q} = \{q_r : \ r=1, \ldots, R\}$$

be a set of “micropanels” $q_r$ that partition the $n$ sample locations into approximately equivalently-sized subsets, such that each $q_r \in \mathcal{Q}$ achieves approximate spatial dispersion across the focal state of each of the $\mathbf{s}_i$ it contains. The $q_r$ defined here are meant to correspond to collections of sample locations referred to in the FIA Program as “Panel 70’s”, which imply $R = 70$, or “Panel 140’s”, which imply $R = 140$.

In CONUS states where the FIA Program is administered by the Pacific Northwest Research Station, $R = 140$ in present day inventory operations. In the rest of the CONUS states, $R = 70$. Though this notation is intended to be flexible enough to accommodate a range of possible values of $R$, in this study we don’t consider PNWRS states, and so we’ll only consider $R = 70$ for practical purposes.

#### Macropanels

The $R$ micropanels can be further aggregated into the set of $M$ macropanels

$$\mathcal{P}_M = \{P_m : \ m = 1,\ldots,M\}$$ for any value of $M$ such that $M \vert R$; that is, such that $R$ is evenly divisible by $M$, and so the macropanels $P_m \in \mathcal{P}_M$ are perfectly evenly partitioned by the micropanels $q_r \in \mathcal{Q}$ (though the micropanels themselves may contain slightly different numbers of $\mathbf{s}_i$.

The macropanels described above are intended to correspond to the \`\`annual panels’’ into which the $s_i$ in each state are partitioned for measurement scheduling purposes. $M$, therefore, is intended to correspond to conceivable state-level inventory cycle lengths prescribed in the FIA program for $R = 70$, and so in this study $M \in \{5, 7, 10, 14\}$. Because the micropanels are initially evenly distributed across the macropanels, the number of micropanels per macropanel can be denoted by $k$ as

$$k = \frac{R}{M}$$

A general expression to recover the identities of the micropanels contained in any macropanel $P_m$ is given by

$$P_m = \{ q_r \in \mathcal{Q} : (m - 1) k \ + \ 1 \leq r \leq m k \}$$.

#### Macropanel Lookup Function

Conversely, to determine the identity of the macropanel to which micropanel $r$ belongs, we can define a macropanel lookup function

$$f(r) = \left\lceil \frac{r}{k} \right\rceil = \left\lceil \frac{r \cdot M}{R} \right\rceil$$ where the ceiling function $\lceil x \rceil$ returns the smallest integer not less than $x$, and $f(r) = m$ outputs the index of the macropanel ($1 \leq m \leq M$) to which the $r^{th}$ micropanel is assigned.
